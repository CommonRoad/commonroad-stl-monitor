```python

from commonroad.common.file_reader import CommonRoadFileReader

scenario_path = "scenarios/test_interstate/DEU_test_safe_distance.xml"
# Open scenario. Note, that the lanelet_assignment parameter has to be True!
scenario,_  = CommonRoadFileReader(scenario_path).open(lanelet_assignment=True)
```

```python
from crmonitor.common.world_state import WorldState

# Create a world state object with ego (the vehicle for which we evaluate)
world = WorldState.create_from_scenario(scenario, ego_obs_id=1003)
```

```python
from crmonitor.evaluation.evaluation import RuleSetEvaluator

# Create a rule set evaluator with defaults, to evaluate all available rules
evaluator = RuleSetEvaluator.create_from_config()
```

```python
# Evaluate the world state object
rule_robustness, predicate_robustness = evaluator.evaluate_incremental(world)
```
